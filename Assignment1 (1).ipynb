```python
#Write a Python program that checks whether a given number is prime or not. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.
```

```python
n = int(input('Enter a number :'))
if n > 1:
    for i in range(2,n):
        if n % i == 0:
            print('Not a prime number')
            break
    else:
        print('Is a prime number')
```

```python
#Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers. The program should then check if the user's answer is correct and display an appropriate message.
```

```python
import random
a = random.randint(1,20)
b = random.randint(1,12)
asked_user = int(input(f'What is the product of {a} and {b}:'))
correct_and = a*b
if asked_user == correct_and:
    print('Correct')
else:
    print('Wrong')
```

```python
#•	Room charges per day: ₹2,000
#•	Doctor consultation fee (flat): ₹1,500
#•	Lab test charges: Based on the number of tests taken. Each test costs ₹300
#•	Medicine charges: Total cost of medicines provided
#•	Discount: If the total bill (before discount) exceeds ₹10,000, a 10% discount is applied
#Create program where cx stayed for 4 days
#had 3 lab tests
#Had medicine charges of 2400
```

```python
room_charges_per_day = 2000
doctor_consultation_fee = 1500
lab_test_charges_per_test = 300
medicine_charges = 2400
total_bill = room_charges_per_day * 4 + doctor_consultation_fee + lab_test_charges_per_test * 3 + medicine_charges
if total_bill > 10000:
  discount = total_bill*0.1

else:
  discount = 0

Final_bill = total_bill - discount
print('Room charges is :', room_charges_per_day * 4)
print('Doctor consultation fee is :', doctor_consultation_fee)
print('Lab test charges is :', lab_test_charges_per_test * 3)
print('Medicine charges is :', medicine_charges)
print('Total bill is :', total_bill)
print('Discount is :', discount)
print('Final bill is :', Final_bill)
```
