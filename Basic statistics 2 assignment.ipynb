```python
#Problem Statement : Hospital Patient Data Analysis
#Context: A hospital maintains patient records including admission details, department, diagnosis, doctor, and bill amount. You have two datasets: one with patient info
#         and another with billing details. Some patients have blank bill amounts, and there are multiple rows for the same patient due to follow-ups.
```

```python
#Tasks:
#           1.	Load the patient dataset and show summary with info().
#           2.	Select only the columns relevant for billing: ['PatientID', 'Department', 'Doctor', 'BillAmount'].
#           3.	Drop administrative columns like ['ReceptionistID', 'CheckInTime'].
#           4.	Use groupby to find total bill amount per department.
#           5.	Remove duplicate patient records based on PatientID.
#           6.	Fill missing BillAmount values with the mean bill amount.
#           7.	Merge the billing dataset with patient dataset on PatientID.
#           8.	Concatenate an additional DataFrame that contains new patients for the current week (row-wise).
#           9.	Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).
```

```python
import pandas as pd
import numpy as np
df_original_patient = pd.read_csv('/content/Patient_Data.csv')
df_original_billing = pd.read_csv('/content/Billing_Data.csv')
```

```python
#2.	Select only the columns relevant for billing: ['PatientID', 'Department', 'Doctor', 'BillAmount'].
df = df_original_patient[['PatientID','Department','Doctor','BillAmount']]
df #select the columns which is mentioned
```

```python
# 4.	Use groupby to find total bill amount per department.
total_bill_per_dept = df.groupby('Department')['BillAmount'].sum()
print('Total bill amount per department:',total_bill_per_dept)
```

```python
# 5.	Remove duplicate patient records based on PatientID.
df_unique = df.drop_duplicates(subset='PatientID',keep='first')
df_unique
```

```python
# 6.	Fill missing BillAmount values with the mean bill amount.
mean_bill = df_unique['BillAmount'].mean()
print(mean_bill)
null_value_filled = df_unique['BillAmount'].fillna(mean_bill)
null_value_filled
```

```python
df_unique['BillAmount'] = null_value_filled
df_unique
```

```python
#7.	Merge the billing dataset with patient dataset on PatientID.
merged_df = pd.merge(df_unique,df_original_billing,on='PatientID')
merged_df
```

```python
# 8.	Concatenate an additional DataFrame that contains new patients for the current week (row-wise).
#Here i am creating my own new patients list because i do not have any other dataframe given
new_patients = {
    'PatientID': [106,107,108],
    'Department':['Neurology','Dermatology','Cardiology'],
    'Doctor':['Dr. John','Dr. Rose','Dr. Smith'],
    'BillAmount':[1000,1500,2000]
}
new_patients_df = pd.DataFrame(new_patients)
new_patients_df
```

```python
concat_df = pd.concat([df_unique,new_patients_df],ignore_index=True)
concat_df
#Without ignore_index=True:
#If df_unique has indices 0-4 and new_patients_df has indices 0-2, the result would have indices 0, 1, 2, 3, 4, 0, 1, 2.

#With ignore_index=True:
#The resulting DataFrame gets a brand new set of indices: 0, 1, 2, 3, 4, 5, 6, 7.
```

```python
# 9.	Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).
```

```python
df_original_billing_final = df_original_billing[['InsuranceCovered', 'FinalAmount']]
new_columns = {
    'InsuranceCovered': [1000,2000,3000],
    'FinalAmount': [1500,2500,3500]
}
new_columns = pd.DataFrame(new_columns)
new_columns
```

```python
concat_columns = pd.concat([df_original_billing_final,new_columns],ignore_index=True)
concat_columns
```

```python
final_df = pd.concat([concat_df,concat_columns],axis=1)
final_df
```
