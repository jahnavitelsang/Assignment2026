```python
#1.	Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.
```

```python
L = [2,8.9,34,1090,2.8794]
D = [2,34,56,78]
L.append(90928)
print('Appended list is :',L)  #Adds item at the last and only one arguement can be passed

L.extend(D)            #adds items from another list or also can pass the numbers directly in the form of list
print('Extended list is :',L)
L.extend([6.7,8.9])
L

L.insert(2,9089098)
print('Inserted list is :' ,L)

L.remove(2)                    # removes 1st matching item passed
print('Removed list is :',L)

L.pop(3)                      # removes and returns the item at the given index , index is passed
print('Popped list is :',L)

L.clear()             #clears everything
print('cleared list :',L)

A = [1,6,2,3,4,5,6]    #finds the index of the item passed
print('indexed item is :',A.index(5))

A.count(6)           #finds how many times number appeared
print('counted item is :',A.count(6))

A.sort()             #sorts in ascending order
print('sorted list is :',A)

A.reverse()            # reverse the list
print('reversed list is :',A)
```
