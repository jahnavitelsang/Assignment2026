```python
import numpy as np
import pandas as pd
from math import sqrt
from scipy.stats import t
from scipy.stats import norm
from scipy.stats import stats
import statsmodels
import warnings
warnings.filterwarnings('ignore')
```

```python
#Data
# A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:
# 1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29
```

```python
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
data = pd.DataFrame(data)
data
```

```python
n = len(data)
n
```

```python
stddata = data.std()
stddata
```

```python
meandata=data.mean()
meandata
```

```python
#a. Build 99% Confidence Interval Using Sample Standard Deviation
```

```python
# As the population standard deviation is not known and the sample size is also small that is just 15 , we use t-distribution.
# Now also calculate the standard of error 1st
```

```python
std_error = stddata/(sqrt(n))
std_error
```

```python
# As the stand of error is calculated , lets find the t values
t_value = t.ppf(0.995, n-1)      # n-1 is the degree of freedom which is df and is calculated as 14 in this case
t_value
```

```python
# now lets calucalte the margin of error
margin_of_error = t_value*std_error
margin_of_error
```

```python
# Now for the confidence interval there is a rangle which is mean+-Margin pf error which means
```

```python
print("99% CI using t-distribution:")
print((meandata - margin_of_error, meandata + margin_of_error))
# the output shows that it is 99% confident that the range lies between 1.090197 - 1.387136
```

```python
#b. Build 99% Confidence Interval Using Known Population Standard Deviation
```

```python
# Since the population std deviation is known that is 0.2 we are using z- distribution .
```

```python
# lets calculate the standard of error .
std_error2 = 0.2 / (sqrt(n))
std_error2
```

```python
# now lets calculate z value
z_value = norm.ppf(0.995)
z_value
```

```python
# The margin of error is
margin_of_error2 = z_value*std_error2
margin_of_error2
```

```python
# Now the CI is
print('CI using z- distribution is :')
print((meandata + margin_of_error, meandata - margin_of_error2))
```

```python
# So now the output lies between  1.106 million and 1.372 million characters with teh std of 0.2
```
